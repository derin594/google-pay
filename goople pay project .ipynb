```python
import pandas as pd

expenses = []

def add_expense():
    name = input("Enter payer name: ")
    amount = float(input("Enter expense amount: "))
    description = input("Enter expense description: ")

    expenses.append({
        "Name": name,
        "Amount": amount,
        "Description": description
    })

    print("Expense added successfully!")


def show_expenses():
    df = pd.DataFrame(expenses)
    print(df)


def calculate_sharing():
    if not expenses:
        print("No expenses available")
        return

    df = pd.DataFrame(expenses)

    total = df["Amount"].sum()
    people = df["Name"].nunique()
    share = total / people

    print("\nTotal Expense:", total)
    print("Number of People:", people)
    print("Share Per Person:", share)

    paid = df.groupby("Name")["Amount"].sum()

    print("\nIndividual Balance:")

    for name, amount in paid.items():
        balance = amount - share

        if balance > 0:
            print(name, "should receive", round(balance, 2))
        elif balance < 0:
            print(name, "should pay", round(abs(balance), 2))
        else:
            print(name, "is settled")


def save_expenses():
    df = pd.DataFrame(expenses)
    df.to_csv("google_pay_expenses.csv", index=False)
    print("Data saved successfully!")


while True:
    print("\n--- GOOGLE PAY SHARING TRACKER ---")
    print("1. Add Expense")
    print("2. View Expenses")
    print("3. Calculate Sharing")
    print("4. Save Expenses")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_expense()
    elif choice == "2":
        show_expenses()
    elif choice == "3":
        calculate_sharing()
    elif choice == "4":
        save_expenses()
    elif choice == "5":
        print("Thank you!")
        break
    else:
        print("Invalid choice")
```
